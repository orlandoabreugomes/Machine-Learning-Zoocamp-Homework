## 2.7 Training Linear Regression: Normal Equation

```python
X = np.array([
    [148, 24, 1385],
    [132, 25, 2031],
    [453, 11, 86],
    [158, 24, 185],
    [172, 25, 201],
    [413, 11, 86],
    [38,  54, 185],
    [142, 25, 431],
    [453, 31, 86]
])

ones = np.ones(X.shape[0])
X = np.column_stack([ones, X])

Y = np.array([10000, 20000, 15000, 20050, 10000, 20000, 15000, 25000, 12000])


Gram_Matrix = X.T.dot(X)
Gram_Matrix_inv = np.linalg.inv(Gram_Matrix)

W = Gram_Matrix_inv.dot(X.T).dot(Y)

print (Gram_Matrix)
print ()
print (Gram_Matrix_inv)
print()
print(Gram_Matrix.dot(Gram_Matrix_inv).round(1))
print()
print(np.linalg.det(Gram_Matrix))
print ()
print(np.array2string(W, suppress_small=True, precision=1))
print ()
print (X.dot(W).round(0))
print()
print((Y * 1.0).round(0))
```

```python
def train_linear_regression(X, Y):

    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])
    
    Gram_Matrix = X.T.dot(X)
    Gram_Matrix_inv = np.linalg.inv(Gram_Matrix)
    W = Gram_Matrix_inv.dot(X.T).dot(Y)

    return W
```

```python
X = np.array([
    [148, 24, 1385],
    [132, 25, 2031],
    [453, 11, 86],
    [158, 24, 185],
    [172, 25, 201],
    [413, 11, 86],
    [38,  54, 185],
    [142, 25, 431],
    [453, 31, 86]
])

W = train_linear_regression(X, Y)
print(np.array2string(W, suppress_small=True, precision=1))
```

```python
def linear_regression(X, W):
    return W[0] + X.dot(W[1: ])
```

```python
Y_prediction = linear_regression(X, W)
print(np.array2string(Y_prediction, suppress_small=True, precision=1))
```

## 2.9 Root Mean Squared Erros (RMSE)

```python
def rmse(y, y_pred):
    error = y_pred - y
    squared_error = error ** 2
    mean_squared_error = squared_error.mean()
    
    return np.sqrt(mean_squared_error)
```

```python
print(rmse(y_train, y_pred).round(5))
```

```python
def prepare_categorical(df, base):
    df = df.copy()
    base = base.copy()
    
    df['age'] = 2017 - df.year
    base += ['age']

    for v in [2, 3, 4]:
        df[f'num_doors_{v}'] = (df.number_of_doors == v).astype(int)
        base += [f'num_doors_{v}']
        
    makes = list(df.make.value_counts().head().index)
    for v in makes:
        df[f'make_{v}'] = (df.make == v).astype(int)
        base += [f'make_{v}']

    X = df[base].fillna(0).values

    return X
```
